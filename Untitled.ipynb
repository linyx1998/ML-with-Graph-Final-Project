```python
def randomize_adjacency_matrix(adjacency_matrix, num_swaps):
    if hasattr(adjacency_matrix, 'toarray'):
        matrix = adjacency_matrix.toarray()
    else:
        matrix = np.array(adjacency_matrix, copy=True)
    
    edges = np.argwhere(np.triu(matrix, k=1) == 1)
    num_edges = len(edges)
    
    for _ in range(num_swaps):
        idx1, idx2 = random.sample(range(num_edges), 2)
        
        u, v = edges[idx1]
        x, y = edges[idx2]
        
        if len({u, v, x, y}) < 4:
            continue
            
        if random.random() > 0.5:
            x, y = y, x
            
        a, b = sorted((u, x))
        c, d = sorted((v, y))
        
        if matrix[a, b] == 0 and matrix[c, d] == 0:
            matrix[u, v] = matrix[v, u] = 0
            matrix[x, y] = matrix[y, x] = 0
            
            matrix[a, b] = matrix[b, a] = 1
            matrix[c, d] = matrix[d, c] = 1
            
            edges[idx1] = [a, b]
            edges[idx2] = [c, d]
            
    return matrix
```

```python
def create_er_graph(adjacency_matrix):
    if hasattr(adjacency_matrix, 'toarray'):
        matrix = adjacency_matrix.toarray()
    else:
        matrix = np.array(adjacency_matrix)

    num_nodes = matrix.shape[0]
    num_edges = np.count_nonzero(np.triu(matrix, k=1))
    
    er_matrix = np.zeros((num_nodes, num_nodes), dtype=int)
    generated_edges = set()
    
    while len(generated_edges) < num_edges:
        needed = num_edges - len(generated_edges)
        batch_size = int(needed * 2) + 100
        
        sources = np.random.randint(0, num_nodes, batch_size)
        targets = np.random.randint(0, num_nodes, batch_size)
        
        mask = sources < targets
        sources = sources[mask]
        targets = targets[mask]
        
        for u, v in zip(sources, targets):
            if len(generated_edges) >= num_edges:
                break
            
            if (u, v) not in generated_edges:
                generated_edges.add((u, v))
                er_matrix[u, v] = 1
                er_matrix[v, u] = 1
                
    return er_matrix
```

```python
def lpca_loss(factors, adj_s, rank): # adj_s = shifted adj with -1's and +1's
    n_row, n_col = adj_s.shape
    U = factors[:n_row*rank].reshape(n_row, rank)
    V = factors[n_row*rank:].reshape(rank, n_col)
    logits = U @ V
    prob_wrong = expit(-logits * adj_s)
    loss = (np.logaddexp(0,-logits*adj_s)).sum()# / n_element    
    U_grad = -(prob_wrong * adj_s) @ V.T# / n_element
    V_grad = -U.T @ (prob_wrong * adj_s)# / n_element
    print(loss)
    return loss, np.concatenate((U_grad.flatten(), V_grad.flatten()))

save_interval = 100 # the number of iterations after which to save the embeddings

def callback_recm(x_i): # prints the loss and periodically saves the factors
    global iter_num
    iter_num += 1
    if iter_num % save_interval == 0:
        global factors, n_row, n_col, rank
        factors = x_i
        U = factors[:n_row*rank].reshape(n_row, rank)
        V = factors[n_row*rank:].reshape(rank, n_col)
        frob_error_norm = np.linalg.norm(clip_01(U@V) - adj) / sp.sparse.linalg.norm(adj)
        print(iter_num, "Frob_error_norm: ", frob_error_norm)
        
# generate a rank k TSVD factorization of a small sparse matrix adj
def factor_TSVD(adj, k):
    w, v = np.linalg.eigh(np.array(adj.todense()))
    order = np.argsort(np.abs(w))[::-1]
    w = w[order[:k]]
    v = v[:,order[:k]]
    U_tsvd, V_tsvd = v * np.sqrt(np.abs(w))[None,:], (np.sign(w)*np.sqrt(np.abs(w)))[:,None] * v.T
    return U_tsvd, V_tsvd

# compute number of triangles in the subgraph induced by the first i nodes
def get_num_tri(i):
    global adj_recon_sort
    sub_adj = adj_recon_sort[:i+1,:i+1]
    num_tri = (1. * sub_adj @ sub_adj @ sub_adj).diagonal().sum() / 6
    return num_tri
```

```python
def lpca_fit(adj, rank_value, maxiter=2000):
    global iter_num, factors, n_row, n_col, rank, frob_error_norm
    rank = rank_value
    n_row, n_col = adj.shape
    factors = -1+2*np.random.random(size=(np.sum(adj.shape)*rank)) # initalize uniformly on [-1,+1]
    iter_num = 0
    U = factors[:n_row*rank].reshape(n_row, rank)
    V = factors[n_row*rank:].reshape(rank, n_col)
    
    res = scipy.optimize.minimize(lpca_loss, x0=factors, 
                              args=(-1 + 2*np.array(adj.todense()), rank), jac=True, method='L-BFGS-B', 
                              callback=callback_recm, 
                              options={'maxiter':maxiter}
                             )
    factors = res.x
    U = res.x[:n_row*rank].reshape(n_row, rank)
    V = res.x[n_row*rank:].reshape(rank, n_col)
    frob_error_norm = np.linalg.norm(1.*(U @ V > 0) - adj) / sp.sparse.linalg.norm(adj)
    print("LPCA Frob norm error: ", frob_error_norm)
    return U, V

def tsvd_fit(adj, rank):
    U_tsvd, V_tsvd = factor_TSVD(adj, rank)
    frob_error_norm = np.linalg.norm(clip_01(U_tsvd@V_tsvd) - adj) / sp.sparse.linalg.norm(adj)
    print("TSVD Frob norm error: ", frob_error_norm)
    return U_tsvd, V_tsvd
```

```python
import numpy as np
import scipy.sparse as sp
import scipy.optimize
from scipy.special import expit

def clip_01(x):
    return np.clip(x, 0, 1)

def lpca_loss(factors, shifted_adj, rank):
    n_row, n_col = shifted_adj.shape
    u_matrix = factors[:n_row * rank].reshape(n_row, rank)
    v_matrix = factors[n_row * rank:].reshape(rank, n_col)
    
    logits = u_matrix @ v_matrix
    prob_wrong = expit(-logits * shifted_adj)
    
    loss = (np.logaddexp(0, -logits * shifted_adj)).sum()
    
    u_grad = -(prob_wrong * shifted_adj) @ v_matrix.T
    v_grad = -u_matrix.T @ (prob_wrong * shifted_adj)
    
    return loss, np.concatenate((u_grad.flatten(), v_grad.flatten()))

def factor_tsvd(adjacency_matrix, k):
    if hasattr(adjacency_matrix, 'todense'):
        dense_matrix = np.array(adjacency_matrix.todense())
    else:
        dense_matrix = np.array(adjacency_matrix)
        
    eigenvalues, eigenvectors = np.linalg.eigh(dense_matrix)
    
    order = np.argsort(np.abs(eigenvalues))[::-1]
    eigenvalues = eigenvalues[order[:k]]
    eigenvectors = eigenvectors[:, order[:k]]
    
    u_tsvd = eigenvectors * np.sqrt(np.abs(eigenvalues))[None, :]
    v_tsvd = (np.sign(eigenvalues) * np.sqrt(np.abs(eigenvalues)))[:, None] * eigenvectors.T
    
    return u_tsvd, v_tsvd

def get_num_tri(matrix, node_limit):
    sub_adj = matrix[:node_limit + 1, :node_limit + 1]
    num_tri = (1. * sub_adj @ sub_adj @ sub_adj).diagonal().sum() / 6
    return num_tri

def lpca_fit(adjacency_matrix, rank_value, max_iter=2000, save_interval=100):
    n_row, n_col = adjacency_matrix.shape
    initial_factors = -1 + 2 * np.random.random(size=(np.sum(adjacency_matrix.shape) * rank))
    
    if hasattr(adjacency_matrix, 'todense'):
        dense_adj = np.array(adjacency_matrix.todense())
    else:
        dense_adj = np.array(adjacency_matrix)
        
    shifted_adj = -1 + 2 * dense_adj
    
    state = {
        'iter_num': 0,
        'current_loss': float('inf')
    }
    
    def objective_wrapper(factors, adj_s, r):
        loss, grad = lpca_loss(factors, adj_s, r)
        state['current_loss'] = loss
        return loss, grad
    
    def callback_wrapper(current_factors):
        state['iter_num'] += 1
        if state['iter_num'] % save_interval == 0:
            u_curr = current_factors[:n_row * rank].reshape(n_row, rank)
            v_curr = current_factors[n_row * rank:].reshape(rank, n_col)
            
            reconstruction = clip_01(u_curr @ v_curr)
            norm_diff = np.linalg.norm(reconstruction - dense_adj)
            norm_adj = sp.linalg.norm(adjacency_matrix) if sp.issparse(adjacency_matrix) else np.linalg.norm(adjacency_matrix)
            
            frob_error = norm_diff / norm_adj
            print(f"Iteration {state['iter_num']}, Loss: {state['current_loss']:.4f}, Frob_error_norm: {frob_error:.6f}")

    res = scipy.optimize.minimize(
        objective_wrapper, 
        x0=initial_factors, 
        args=(shifted_adj, rank), 
        jac=True, 
        method='L-BFGS-B', 
        callback=callback_wrapper, 
        options={'maxiter': max_iter}
    )
    
    final_u = res.x[:n_row * rank].reshape(n_row, rank)
    final_v = res.x[n_row * rank:].reshape(rank, n_col)
    
    binary_recon = 1. * ((final_u @ final_v) > 0)
    norm_adj = sp.linalg.norm(adjacency_matrix) if sp.issparse(adjacency_matrix) else np.linalg.norm(adjacency_matrix)
    final_error = np.linalg.norm(binary_recon - dense_adj) / norm_adj
    
    print("LPCA Final Frob norm error: ", final_error)
    return final_u, final_v

def tsvd_fit(adjacency_matrix, rank):
    u_tsvd, v_tsvd = factor_tsvd(adjacency_matrix, rank)
    
    if hasattr(adjacency_matrix, 'todense'):
        dense_adj = np.array(adjacency_matrix.todense())
    else:
        dense_adj = np.array(adjacency_matrix)
        
    reconstruction = clip_01(u_tsvd @ v_tsvd)
    norm_diff = np.linalg.norm(reconstruction - dense_adj)
    norm_adj = sp.linalg.norm(adjacency_matrix) if sp.issparse(adjacency_matrix) else np.linalg.norm(adjacency_matrix)
    
    frob_error_norm = norm_diff / norm_adj
    print("TSVD Frob norm error: ", frob_error_norm)
    return u_tsvd, v_tsvd
```
